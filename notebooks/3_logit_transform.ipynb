# Logistic regression: from probabilities to log-odds

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

from scripts.logit_transform_viz import plot_logit_morph
```

## Data and model

```python
rng = np.random.default_rng(42)
n = 80

hours = rng.uniform(0, 10, n)
p_true = 1 / (1 + np.exp(-0.8 * (hours - 5)))
passed = (rng.random(n) < p_true).astype(float)

model = LogisticRegression(C=np.inf, max_iter=10_000).fit(hours.reshape(-1, 1), passed)
w, b = model.coef_[0, 0], model.intercept_[0]
print(f"w={w:.3f}, b={b:.3f}")
```

## Log-odds transform

```python
plot_logit_morph(hours, passed, w, b).show()
```
